# **Import Libraries**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

print("Libraries loaded successfully.")
```

# **Load Dataset and Find Missing/Null Values**

```python
# Load dataset
df = pd.read_csv("/content/MNC_IT_employee_attrition_sample.csv")

# Display first 5 rows
print("First 5 rows:")
display(df.head())

# Display dataset information
print("\nDataset Information:")
df.info()

# Display descriptive statistics
print("\nDescriptive Statistics:")
display(df.describe())

# Check missing / null values
print("\nMissing / Null Values:")
print(df.isnull().sum())

# Check whether any missing values exist
print("\nAny missing values:", df.isnull().values.any())
```

# **Data Cleaning and Filling Missing Values**

```python
# Remove duplicate rows
df.drop_duplicates(inplace=True)

# Clean Employee_ID
df['Employee_ID'] = df['Employee_ID'].astype(str).str.strip()

# Remove duplicate Employee IDs
df.drop_duplicates(subset=['Employee_ID'], keep='first', inplace=True)

# Identify categorical and numerical columns
categorical_cols = df.select_dtypes(include=['object']).columns
numerical_cols = df.select_dtypes(include=[np.number]).columns

# Fill missing values in categorical columns
for col in categorical_cols:
    df[col] = df[col].fillna('Unknown').str.strip()

# Fill missing values in numerical columns using median
for col in numerical_cols:
    df[col] = df[col].fillna(df[col].median())

# Check missing values after cleaning
print("Missing values after cleaning:")
print(df.isnull().sum())

print("\nAny missing values after cleaning:",
      df.isnull().values.any())

print("\nCleaned Shape:", df.shape)
```

# **Data Transformation and Encoding**

```python
# Remove extra spaces from column names
df.columns = df.columns.str.strip()

# Encode Job Level
if 'Job_Level' in df.columns:
    df['Job_Level_Encoded'] = (
        df['Job_Level']
        .str.replace('L', '', regex=False)
        .astype(int)
    )

# Encode Salary Growth Potential
if 'Salary_Growth_Potential' in df.columns:
    df['Salary_Growth_Encoded'] = df[
        'Salary_Growth_Potential'
    ].map({
        'Low': 0,
        'Medium': 1,
        'High': 2
    })

# Encode Attrition
if 'Attrition' in df.columns:
    df['Attrition_Binary'] = df['Attrition'].map({
        'No': 0,
        'Yes': 1
    })

# Remove original categorical columns after encoding
df.drop(
    columns=[
        'Job_Level',
        'Salary_Growth_Potential',
        'Attrition'
    ],
    errors='ignore',
    inplace=True
)

# Nominal categorical features
nominal_features = [
    'Company',
    'Department',
    'Job_Role',
    'Location',
    'Gender',
    'Education',
    'Employment_Type',
    'Work_Mode'
]

# Keep only columns that exist
nominal_features = [
    col for col in nominal_features
    if col in df.columns
]

# One-hot encoding
df_transformed = pd.get_dummies(
    df,
    columns=nominal_features,
    drop_first=True,
    dtype=int
)

print(
    "Data transformation completed."
)

print(
    "Transformed Shape:",
    df_transformed.shape
)

display(df_transformed.head())
```

# **Data Preprocessing / Standardization**

```python
from sklearn.preprocessing import StandardScaler

# Numerical columns to scale
columns_to_scale = [
    'Age',
    'Years_at_Company',
    'Years_in_Current_Role',
    'Total_Work_Experience',
    'Monthly_Salary_INR',
    'Annual_Salary_INR',
    'Annual_Bonus_INR',
    'Job_Satisfaction',
    'Work_Life_Balance',
    'Performance_Rating',
    'Overtime_Hours_Month',
    'Training_Hours_Year',
    'Projects_Count',
    'Promotions_Last_5_Years',
    'Manager_Rating',
    'Distance_From_Home_KM',
    'Job_Level_Encoded',
    'Salary_Growth_Encoded'
]

# Keep only columns that exist
columns_to_scale = [
    col for col in columns_to_scale
    if col in df_transformed.columns
]

# Create scaler
scaler = StandardScaler()

# Apply standardization
df_transformed[columns_to_scale] = scaler.fit_transform(
    df_transformed[columns_to_scale]
)

print("Numerical features have been standardized.")

display(
    df_transformed[columns_to_scale].head()
)
```

# **Mean, Median and Mode**

```python
mean_monthly_salary = df["Monthly_Salary_INR"].mean()
median_monthly_salary = df["Monthly_Salary_INR"].median()
mode_monthly_salary = df["Monthly_Salary_INR"].mode()[0]
print(f"Mean Monthly Salary:   {mean_monthly_salary:.2f}")
print(f"Median Monthly Salary: {median_monthly_salary:.2f}")
print(f"Mode Monthly Salary:   {mode_monthly_salary:.2f}")
```

# **Variance, Standard Deviation and CV**

```python
# Variance
variance_monthly_salary = (df["Monthly_Salary_INR"].var())

# Standard deviation
std_dev_monthly_salary = (df["Monthly_Salary_INR"].std())

# Coefficient of Variation
cv_monthly_salary = (std_dev_monthly_salary / mean_monthly_salary) * 100

print(f"Variance Monthly Salary: "f"{variance_monthly_salary:.2f}")
print(f"Standard Deviation Monthly Salary: "f"{std_dev_monthly_salary:.2f}")
print(f"Coefficient of Variation (CV): "f"{cv_monthly_salary:.2f}%")
```
